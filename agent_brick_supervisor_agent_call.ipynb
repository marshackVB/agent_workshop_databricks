### Call the Agent Bricks supervisor agent API

```python
supervisor_url = ""

# Parse base_url and model from the supervisor_url
parts = supervisor_url.rsplit("/", 2)
base_url = parts[0]
model_id = parts[1]
```

```python
from openai import OpenAI
import os

DATABRICKS_TOKEN = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()

client = OpenAI(
    api_key=DATABRICKS_TOKEN,
    base_url=base_url
)

response = client.responses.create(
    model=model_id,
    input=[
        {
            "role": "user",
            "content": "What are our top 10 best selling products?"
        }
    ]
)

# Extract only the final message (last message output), excluding intermediate thinking and tool calls
final_output = [o for o in response.output if o.type == "message"][-1]
final_text = " ".join(getattr(c, "text", "") for c in final_output.content)
print(final_text)
```
